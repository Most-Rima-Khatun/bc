```python
import hashlib
```

```python
string = 'Hello , world !'
hash_object = hashlib.sha256()
```

```python
hash_object.update(string.encode('utf-8'))
```

```python
hex_dig = hash_object.hexdigest()
print('String:{}'.format(string))
```

```python
print('hash value sha-256:{}'.format(hex_dig))
```
